```python
import pandas as pd

sales = {
    "month": ["Jan", "Feb", "Mar", "Apr", "May"],
    "sales": [
        12000000,
        14500000,
        16200000,
        15800000,
        18100000
    ]
}

#df1 = pd.DataFrame(sales)

data = {
    "product": ["Speaker", "TV", "Fan", "Aircon", "Speaker"],
    "price": [450000, 1200000, 150000, 1800000, 550000],
    "quantity": [2, 1, 5, 1, 3]
}

df2 = pd.DataFrame(data)

# Filter products with price > 500000
df_filtered = df2[df2['price'] > 500000].copy()

# Add `total` column = price * quantity
df_filtered['total'] = df_filtered['price'] * df_filtered['quantity']

# Compute total sales
total_sales = df_filtered['total'].sum()

# Show results
print('Filtered products (price > 500000):')
print(df_filtered)

print('\nTotal sales for filtered products:', total_sales)
```
