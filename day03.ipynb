```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 1. NumPy basics
sales = np.array([
    12000000,
    14500000,
    16200000,
    15800000,
    18100000
])

print(sales)
print("Mean:", np.mean(sales))
print("Maximum:", np.max(sales))
print("Minimum:", np.min(sales))
print("Total:", np.sum(sales))

# 2. Python list vs NumPy array
prices = [100, 200, 300]
print("Python list * 2:", prices * 2)

prices_np = np.array([100, 200, 300])
print("NumPy array * 2:", prices_np * 2)

# 3. Vectorized operation
discounted_prices = np.array([
    450000,
    1200000,
    150000,
    1800000,
    550000
]) * 0.9

print("Discounted prices:", discounted_prices)

# 4. Feature vector / matrix
product = np.array([450000, 2, 1])

products = np.array([
    [450000, 2, 1],
    [1200000, 1, 0],
    [150000, 5, 1],
    [1800000, 1, 0],
    [550000, 3, 1]
])

print("Product vector:", product)
print("Products matrix:\n", products)
print("Shape:", products.shape)

# 5. Data visualization with pandas + matplotlib
data = {
    "month": ["Jan", "Feb", "Mar", "Apr", "May"],
    "sales": [12000000, 14500000, 16200000, 15800000, 18100000]
}

df = pd.DataFrame(data)

plt.plot(df["month"], df["sales"])
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthly Sales")
plt.show()

# 6. Bar chart
product_names = ["Speaker", "TV", "Fan", "Aircon"]
product_sales = [25, 10, 40, 8]

plt.bar(product_names, product_sales)
plt.xlabel("Product")
plt.ylabel("Units Sold")
plt.title("Product Sales")
plt.show()

# 7. Day 3 exercise dataset
exercise_data = {
    "advertising": [100000, 200000, 300000, 400000, 500000],
    "sales": [1200000, 1800000, 2500000, 3100000, 3900000]
}

exercise_df = pd.DataFrame(exercise_data)
print(exercise_df)

print("Advertising average:", exercise_df["advertising"].mean())
print("Sales average:", exercise_df["sales"].mean())

advertising_array = exercise_df["advertising"].to_numpy()
sales_array = exercise_df["sales"].to_numpy()

print("Advertising array:", advertising_array)
print("Sales array:", sales_array)
print("Advertising shape:", advertising_array.shape)
print("Sales shape:", sales_array.shape)

plt.plot(exercise_df["advertising"], exercise_df["sales"])
plt.xlabel("Advertising")
plt.ylabel("Sales")
plt.title("Advertising vs Sales")
plt.show()
```
