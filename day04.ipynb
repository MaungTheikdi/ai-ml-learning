```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
```

```python
advertising = np.array([100, 200, 300, 400, 500])
sales = np.array([1.2, 1.9, 2.5, 3.2, 3.9])

print("Advertising:", advertising)
print("Sales:", sales)
```

```python
X = advertising.reshape(-1, 1)
y = sales

print("X:")
print(X)

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
model = LinearRegression()
model.fit(X, y)

print("Model training completed.")
```

```python
new_advertising = np.array([[350]])

predicted_sales = model.predict(new_advertising)

print("Predicted sales:", predicted_sales[0], "million MMK")
```

```python
predicted_values = model.predict(X)

plt.scatter(
    advertising,
    sales,
    color="blue",
    label="Actual Sales"
)

plt.plot(
    advertising,
    predicted_values,
    color="red",
    label="Prediction Line"
)

plt.xlabel("Advertising Budget")
plt.ylabel("Sales (Million MMK)")
plt.title("Advertising vs Sales")
plt.legend()
plt.grid(True)
plt.show()
```

```python
print("Coefficient:", model.coef_[0])
print("Intercept:", model.intercept_)
```

```python
budgets = np.array([
    [150],
    [250],
    [350],
    [450],
    [600]
])

predictions = model.predict(budgets)

for budget, prediction in zip(budgets, predictions):
    print(
        f"Advertising: {budget[0]} သိန်း → "
        f"Predicted Sales: {prediction:.2f} သန်း"
    )
```
